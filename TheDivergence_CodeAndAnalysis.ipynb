# Analysis & Explainer

This section documents the analytical and design choices behind *The Divergence* interactive narrative.

## 1. Motivation and Dataset

### 1.1 Motivation

The motivation for this project is rooted in the Resource Curse paradox. This is a phenomenon where countries with an abundance of natural resources often experience lower economic growth and worse development outcomes than countries with fewer natural resources (Sachs & Warner, 1995). By performing a comparative analysis between South Korea and Nigeria, we explore a historical Natural Experiment. In the early 1960s, both nations shared similar profiles of post-colonial poverty and comparable GDP per capita. Today, they represent opposite ends of the developmental spectrum.

This project seeks to challenge the deterministic view that natural wealth equates to prosperity. While Nigeria leveraged Black Gold in the form of oil, South Korea invested in Human Capital and industrial complexity. Our narrative aims to spark a reflection on the role of institutional policy and resource management in shaping a nation's destiny.

### 1.2 What is your dataset?

To tell this story accurately, we utilized a multi-dimensional data approach. The primary backbone is the Gapminder Dataset, which provides historical data for GDP per capita (PPP), Life Expectancy, and Population from 1950 to the present. We also utilized the World Bank Open Data to extract structural indicators such as Manufacturing Value Added as a percentage of GDP and School Enrollment to explain the specific drivers behind the growth. Furthermore, we incorporated data from the Observatory of Economic Complexity (OEC). This was crucial for analyzing the shift in export baskets, specifically moving from raw materials to high-tech goods in Korea versus the continued oil-dependency in Nigeria. Finally, we used the BP Statistical Review and Transparency International data to provide the necessary context for oil shocks and the Corruption Perceptions Index. These help visualize the impact of external shocks and internal governance.

| Dataset | Source | Variables used |
|---|---|---|
| Gapminder | [gapminder.org](https://www.gapminder.org/data/) | GDP per capita (PPP), Life Expectancy, Population, Fertility Rate (1952–2007) |
| World Bank Open Data | [data.worldbank.org](https://data.worldbank.org) | Manufacturing Value Added (% GDP), Secondary School Enrollment by gender |
| Observatory of Economic Complexity (OEC) | [oec.world](https://oec.world) | Export basket composition — shift from raw materials to manufactures |
| BP Statistical Review | [bp.com](https://www.bp.com/en/global/corporate/energy-economics/statistical-review-of-world-energy.html) | Oil production and price shocks context |
| Transparency International | [transparency.org](https://www.transparency.org) | Corruption Perceptions Index — governance context |

### 1.3 What was your goal for the end user's experience?

The goal of this project is to guide the user through a structured cognitive journey that moves from simple observation to critical evaluation. Initially, the user is presented with interactive scatter plots where they can observe the starting parity between the two nations. This phase establishes the baseline of the story, showing that the massive divergence we see today was not inevitable.

As the user progresses through the chapter system, the experience shifts toward active exploration. By interacting with specific variables such as education levels or industrial shifts, the user can investigate the hidden drivers of the divergence. The dialogue between the characters Ji-ho and Emeka is designed to facilitate this by translating complex economic data into a relatable human story. This reduces the cognitive load and keeps the user engaged with the narrative.

In the final phase, the goal is for the user to synthesize these insights. By reflecting on the differences in resilience and resource dependency, the user reaches a nuanced conclusion about development. The interactive nature of the story ensures that the user is not just a passive reader but an active participant in uncovering how policy and management choices shaped the modern history of these two nations.

## 2. Basic Stats — Data Cleaning and EDA

### 2.1 Data Cleaning and Preprocessing

The primary data source is the Gapminder Python package, providing five year interval observations from 1952 to 2007 for 142 countries. We supplemented this with World Bank data on school enrollment and manufactured exports. The most critical cleaning task was building a manual 85 country lookup table to align Gapminder naming conventions with ISO 3166-1 alpha-3 codes. This was essential for the Plotly choropleth renderer to function correctly.

For the export composition analysis, we manually encoded data into four categories for six benchmark years. We chose a discrete snapshot approach rather than interpolation to emphasize structural shifts over time. Additionally, we capped GDP per capita at the 97th continent percentile for choropleth animations. This prevents outliers from compressing the color scale, ensuring the visual differences between Nigeria and Korea remains distinct and readable.

### 2.2 Exploratory Data Analysis

Our analysis shows a sharp divergence between 1967 and 1977. While both nations started at parity in the early 1950s, this decade marked the split where Korea's industrial drive began and Nigeria's oil dependency solidified. Statistically, Nigeria's growth is 3.9 times more volatile than Korea's, indicating that wealth arrived in spikes rather than steady accumulation.

The export data reveals a total structural reversal. By 2007, Korea's basket was over 90 percent manufactures, while Nigeria's was over 90 percent fuel, proving that oil revenue actively crowded out other sectors in Nigeria. Finally, Korea is a clear demographic outlier. In the 1970s and 1980s, Korea's secondary school enrollment sat two standard deviations above the median for its income level, while Nigeria merely tracked the average regression line. ## 3. Data Analysis and Machine Learning

### 3.1 Data Analysis

The analysis is entirely descriptive and comparative. The goal was to build a causal narrative from visual evidence rather than to forecast outcomes.

**Chapter 1 — The Divergence.** Plotting both countries on the same GDP axis from 1952 to 2007 shows the gap isn't new; it opened in the early 1970s and never stopped growing. To prove Nigeria's growth was fragile, I calculated that its growth volatility is 4 times higher than Korea's. This single number anchors the argument that Nigeria's wealth arrived in spikes rather than steady accumulation.

**Chapter 2 — The Structural Trap.** Stacked bar charts show Korea's export mix climbed the value chain from agriculture to electronics, while Nigeria's simplified into 90 percent fuel dependency. This is a clear case of Dutch Disease (Corden & Neary, 1982), where oil revenue crowded out other sectors. The data confirms this because fuel and manufacturing shares are near perfect mirror images over time.

**Chapter 3 — The Demographic Dividend.** A connected scatter plot of fertility vs. life expectancy shows the demographic transition. Korea went from 6 children/woman to 1.5, while Nigeria's path is much shallower. Also, Korea's school enrollment was way above the average for its income level in the 70s, suggesting education investment happened before the economic boom. The choropleth maps show Korea as an outlier in Asia, while Nigeria barely shifted compared to resource poor neighbors like Ethiopia.

### 3.2 Machine Learning

No machine learning was used. With only 12 observations per country (every 5 years), the data is way too small for supervised models. The questions I'm asking are about why and when things happened, which is about context, not prediction. Trying to run a model here would be statistically meaningless and wouldn't add any real insight. Everything in the project is about aggregation and visual encoding.

## 4. Genre, Visual Narrative, and Narrative Structure

### 4.1 Genre: The Comic Strip

According to the typology by Segal and Heer (2010), "The Divergence" uses the Comic Strip genre. It features character panels (Ji-ho on the left, Emeka on the right), sequential click-to-advance pages, and shifting portrait expressions that reflect the emotional tone of the data. Unlike a standalone graphic, the data charts are embedded directly within the character dialogue frames. This digital hybrid format ensures that the characters drive the narrative, making the data an integrated part of their conversation.

### 4.2 Visual Narrative Tools (Figure 7 — Segal and Heer, 2010)

We utilized all three categories of visual narrative tools to guide the user:

#### Visual Structuring

|---|---|

#### Transition Guidance
| Tool | Implementation |
|---|---|
| **Object Constancy** | Is maintained because the chart container stays fixed while the data layer updates based on user choices |
| **Typewriter Effect** | On text guides reading pace |
| **Continuity Editing** | Via the "continue" indicators ensures a clear temporal flow between narrative beats |

---

### 4.3 Narrative Structure Tools (Figure 7 — Segal and Heer, 2010)

#### Ordering
We used a Martini Glass structure. The stem consists of the linear prologue and chapter openings that provide essential context. The bowl is the User-Directed portion where readers choose between A, B, or C paths to explore different data dimensions before converging back to a common bridge.

#### Interactivity
| Tool | Implementation |
|---|---|
| **Navigation Buttons** | Ensure the reader advances the story actively |
| **Filtering / Selection** | Occurs when users choose an A/B/C path, which physically changes the data layer shown |
| **Prediction Prompts** | As a "tacit tutorial" to create a cognitive stake in the data before it is revealed |
| **Explicit Instructions** | In the prompt fields help reduce user friction |

#### Messaging
| Tool | Implementation |
|---|---|
| **Annotations** | Synchronized with character dialogue to link verbal and visual points |
| **Chapter Title Cards** | Reorient the reader at each new phase |
| **Captions** | So the data is labeled clearly |
| **Repetition** | Of the chapter structure to focus the reader's attention on the economic argument rather than the interface |

## 5. Visualizations

### 5.1 Visualization 1 — GDP Line Chart *(Chapter 1)*

**What it is:** A dual line time series from 1952 to 2007 with one line per country and two optional overlays. Option A provides a shaded critical window with event annotations, while Option B adds a growth rate bar panel beneath the main chart.

**Why it is right for this story:** The story's central claim is that two lines starting at the same point ended twenty-one thousand dollars apart. A line chart makes this claim self evident before the user reads a single word. You do not explain the divergence; you simply reveal it. Option A is where the argument lives, showing that while both nations occupied the same window, their conditions were worlds apart. Option B uses a second panel of bars to show structural cause, specifically Korea's consistent stability versus Nigeria's swings that track oil prices.

---

### 5.2 Visualization 2 — Export Composition Stacked Bars *(Chapter 2)*

**What it is:** Side by side stacked bar charts showing export composition across decades with optional small multiples, a geopolitical event timeline, and a single 2007 snapshot.

**Why it is right for this story:** This chapter is about structure and what these countries were actually building. A stacked bar is the perfect encoding for composition. The visual metaphors of Korea as a staircase climbing from textiles to semiconductors and Nigeria as a wall locked in 96 percent oil dependency since 1982 are made literal here. Small multiples are used so the reader does not have to rely on memory. By scanning the panels, the total absence of change in Nigeria's export structure becomes visually unmistakable.

---

### 5.3 Visualization 3 — Connected Scatter Plot *(Chapter 3)*

**What it is:** Fertility rate vs. life expectancy with each country's path drawn as a connected line of dots from 1952 to 2007. Options include a gender enrollment dumbbell, a 23 country secondary school enrollment dot plot, and animated continental choropleth maps.

**Why it is right for this story:** The demographic transition is a two dimensional shift. A connected scatter plot is the only way to encode movement through two variable space as a visible path. The geometry of the paths tells the story. Korea's sweeps toward the wealthy nation corner, while Nigeria's climbs a vertical wall of increasing life expectancy with no fertility change. Option B shifts the scale, showing Korea is now adjacent to Scandinavia while Nigeria remains at the bottom of a global cluster. Finally, the animated choropleths broaden the zoom to entire continents, proving that Korea and Nigeria are representative of their respective regional patterns.

## 6. Discussion

### 6.1 Critical Thinking

The final structure of the webpage was a deliberate evolution. I initially started building the project with a visual novel video game structure but realized that many of those mechanics did not help the user understand the data correctly. However, I kept several elements like the user options, the page design, and the character portraits which give the visualization a lot of originality. The core story of comparing two countries from different regions that started at the same point is very strong. Seeing the massive difference between them today provides a powerful and interesting insight into economic development.

I am particularly happy with the design of the plots. They are clean and well designed in terms of colors and typography. Even though some are relatively simple, they deliver clear insights. I also believe I used AI properly throughout the process to ensure technical accuracy and to find interesting narrative paths. The project changed significantly from its initial concept to avoid being overwhelming, and I decided to make the characters speak in a more article like way. I found that if characters interrupt each other too much, it makes the data harder to understand.

### 6.2 Limitations and Future Improvements

There are certainly areas that could be improved. While the reasons for the divergence are clear, it was sometimes hard to find shocking new insights because the data is very explanatory. A unexpected twist would have been a nice addition. I also think more complex plot types could have been used, especially with the map data, but since the focus was strictly on comparing two specific countries, adding more nations often felt like it would break the narrative point. This led to some of the plots feeling a bit simple.

The dialogue is currently very explanatory and could be improved to be less boring. It goes straight to the point, but I could have added more introductory hooks to get the user more interested in the topic. I also would have liked to add pictures to support what the characters were saying, but time limitations prevented this. Finally, the user interaction could have been more meaningful. Instead of just asking questions that do not change the outcome, I wanted to create a system where the user chooses between the characters' viewpoints, leading to a final page that reflects their choices. I focused my time on the actual data story, the design, and the insights rather than these more complex video game mechanics.

### CH1 Option B — Volatility & Crisis Comparison

```python
def annualised_growth(series):
    """Compute annualised % growth rate for each 5-year gapminder interval."""
    yrs = sorted(series.index)
    return pd.Series(
        {yrs[i]: ((series[yrs[i]] / series[yrs[i-1]]) ** (1/(yrs[i]-yrs[i-1])) - 1) * 100
         for i in range(1, len(yrs))})

growth_korea   = annualised_growth(kor['gdpPercap'])
growth_nigeria = annualised_growth(nig['gdpPercap'])
vol_ratio = growth_nigeria.std() / growth_korea.std()

fig, (ax1, ax2) = plt.subplots(2, 1, figsize=(12,9),
                                gridspec_kw={'height_ratios':[2,1.5]},
                                sharex=True)
fig.patch.set_facecolor(BG)
for ax in (ax1, ax2): style_ax(ax)

# TOP PANEL
ax1.plot(kor.index, kor['gdpPercap'], color=C_KOREA,   lw=2.5, marker='o', ms=5, label='South Korea')
ax1.plot(nig.index, nig['gdpPercap'], color=C_NIGERIA, lw=2.5, marker='o', ms=5, label='Nigeria')
ax1.axvspan(1980, 1987, color='red', alpha=0.07, label='Oil price crash 1980-87')
ax1.axvline(1997, color='#666666', linestyle='--', lw=1.2, label='Asian Financial Crisis')
ax1.yaxis.set_major_formatter(FuncFormatter(lambda x,_: f'${x:,.0f}'))
ax1.set_title('GDP per Capita', fontsize=12, pad=8)
ax1.legend(loc='upper left', framealpha=0.9, edgecolor=GRID_CLR, fontsize=9)

# BOTTOM PANEL — annualised growth bars
years = kor.index.tolist()
x_kor = np.array(years) - 0.2
x_nig = np.array(years) + 0.2
ax2.bar(x_kor, growth_korea.reindex(years),   width=0.4, color=C_KOREA,
        alpha=0.9, edgecolor='white', lw=0.5, label='South Korea')
ax2.bar(x_nig, growth_nigeria.reindex(years), width=0.4, color=C_NIGERIA,
        alpha=0.9, edgecolor='white', lw=0.5, label='Nigeria')
ax2.axhline(0, color='#aaaaaa', lw=1)
ax2.axvspan(1980, 1987, color='red', alpha=0.07)
ax2.axvline(1997, color='#666666', linestyle='--', lw=1.2)
ax2.set_title('Annualised GDP Growth Rate (% p.a., 5-yr avg)', fontsize=12, pad=8)
ax2.legend(loc='upper left', framealpha=0.9, edgecolor=GRID_CLR, fontsize=9)
ax2.set_xlim(1950, 2009)
ax2.set_xticks(range(1952, 2008, 10))
ax2.yaxis.set_major_formatter(mticker.FuncFormatter(lambda x,_: f'{x:.1f}%'))

# volatility annotation
ax2.text(0.98, 0.98,
         f"Nigeria growth volatility: {vol_ratio:.1f}x higher than Korea's\n(std. dev. of annualised 5-yr growth rates)",
         transform=ax2.transAxes, ha='right', va='top', fontsize=9.5,
         bbox=dict(boxstyle='round,pad=0.4', facecolor='white', edgecolor=GRID_CLR, alpha=0.95))

fig.suptitle("Option B - The Volatility Gap: Korea's Steady Growth vs Nigeria's Boom-Bust Cycles",
             fontsize=13, y=0.995)
ax2.text(1.0, -0.12, 'Source: Gapminder', transform=ax2.transAxes,
         ha='right', va='top', fontsize=8.5, color='#999999', style='italic')
plt.tight_layout()
plt.savefig('figures/ch1_optionB.png', dpi=180, bbox_inches='tight')
plt.show()
print(f"Saved ch1_optionB.png  |  vol_ratio={vol_ratio:.2f}")
```

```python
import os, sys

import pandas as pd
import plotly.express as px
from gapminder import gapminder as gap

BG = '#f9f9f9'

ISO_MAP = {
    # ── Africa (all 52 gapminder countries) ──────────────────────────────────
    'Algeria':                  'DZA', 'Angola':               'AGO',
    'Benin':                    'BEN', 'Botswana':             'BWA',
    'Burkina Faso':             'BFA', 'Burundi':              'BDI',
    'Cameroon':                 'CMR', 'Central African Republic': 'CAF',
    'Chad':                     'TCD', 'Comoros':              'COM',
    'Congo, Dem. Rep.':         'COD', 'Congo, Rep.':          'COG',
    "Cote d'Ivoire":            'CIV', 'Djibouti':             'DJI',
    'Egypt':                    'EGY', 'Egypt, Arab Rep.':     'EGY',
    'Equatorial Guinea':        'GNQ', 'Eritrea':              'ERI',
    'Ethiopia':                 'ETH', 'Gabon':                'GAB',
    'Gambia':                   'GMB', 'Ghana':                'GHA',
    'Guinea':                   'GIN', 'Guinea-Bissau':        'GNB',
    'Kenya':                    'KEN', 'Lesotho':              'LSO',
    'Liberia':                  'LBR', 'Libya':                'LBY',
    'Madagascar':               'MDG', 'Malawi':               'MWI',
    'Mali':                     'MLI', 'Mauritania':           'MRT',
    'Mauritius':                'MUS', 'Morocco':              'MAR',
    'Mozambique':               'MOZ', 'Namibia':              'NAM',
    'Niger':                    'NER', 'Nigeria':              'NGA',
    'Reunion':                  'REU', 'Rwanda':               'RWA',
    'Sao Tome and Principe':    'STP', 'Senegal':              'SEN',
    'Sierra Leone':             'SLE', 'Somalia':              'SOM',
    'South Africa':             'ZAF', 'Sudan':                'SDN',
    'Swaziland':                'SWZ', 'Tanzania':             'TZA',
    'Togo':                     'TGO', 'Tunisia':              'TUN',
    'Uganda':                   'UGA', 'Zambia':               'ZMB',
    'Zimbabwe':                 'ZWE',
    # ── Asia (all 33 gapminder countries) ────────────────────────────────────
    'Afghanistan':              'AFG', 'Bahrain':              'BHR',
    'Bangladesh':               'BGD', 'Cambodia':             'KHM',
    'China':                    'CHN', 'Hong Kong, China':     'HKG',
    'India':                    'IND', 'Indonesia':            'IDN',
    'Iran':                     'IRN', 'Iran, Islamic Rep.':   'IRN',
    'Iraq':                     'IRQ', 'Israel':               'ISR',
    'Japan':                    'JPN', 'Jordan':               'JOR',
    'Korea, Dem. Rep.':         'PRK', 'Korea, Rep.':          'KOR',
    'Kuwait':                   'KWT', 'Lebanon':              'LBN',
    'Malaysia':                 'MYS', 'Mongolia':             'MNG',
    'Myanmar':                  'MMR', 'Nepal':                'NPL',
    'Oman':                     'OMN', 'Pakistan':             'PAK',
    'Philippines':              'PHL', 'Saudi Arabia':         'SAU',
    'Singapore':                'SGP', 'Sri Lanka':            'LKA',
    'Syria':                    'SYR', 'Taiwan':               'TWN',
    'Thailand':                 'THA', 'Vietnam':              'VNM',
    'West Bank and Gaza':       'PSE', 'Yemen':                'YEM',
    'Yemen, Rep.':              'YEM',
}

gap_full = gap.copy()
gap_full['iso'] = gap_full['country'].map(ISO_MAP)
# Make year a string so animation labels read "1952", "1957", etc.
gap_full['year_str'] = gap_full['year'].astype(str)

africa_df = gap_full[gap_full['continent'] == 'Africa'].copy().sort_values('year')
asia_df   = gap_full[gap_full['continent'] == 'Asia'].copy().sort_values('year')

def build_gdp_map(df, scope, title, note, fname):
    # Clip at 97th pct so one outlier doesn't bleach everything
    cap = float(df['gdpPercap'].quantile(0.97))

    fig = px.choropleth(
        df,
        locations='iso',
        color='gdpPercap',
        animation_frame='year_str',
        animation_group='country',
        color_continuous_scale='YlOrRd',
        range_color=[0, cap],
        scope=scope,
        labels={
            'gdpPercap': 'GDP per Capita (USD)',
            'year_str':  'Year',
            'iso':       '',
        },
        hover_name='country',
        hover_data={
            'gdpPercap': ':,.0f',
            'year_str':  True,
            'iso':       False,
        },
    )

    fig.update_geos(
        projection_type='natural earth',
        showframe=False,
        showcoastlines=True,
        coastlinecolor='#cccccc',
        showland=True,
        landcolor='#eeeeee',
        showocean=True,
        oceancolor='#d4e8f5',
        bgcolor=BG,
    )

    fig.update_layout(
        title=dict(
            text=f'<b>{title}</b>',
            x=0.5, xanchor='center',
            y=0.97,
            font=dict(size=14, color='#222'),
        ),
        paper_bgcolor=BG,
        plot_bgcolor=BG,
        height=580,
        margin=dict(l=0, r=0, t=55, b=50),
        coloraxis_colorbar=dict(
            title=dict(text='GDP per<br>Capita (USD)', font=dict(size=10)),
            tickformat='$,.0f',
            tickfont=dict(size=9),
            len=0.75,
            tickvals=[0, cap * 0.25, cap * 0.5, cap * 0.75, cap],
            ticktext=[
                '$0',
                f'${cap*0.25:,.0f}',
                f'${cap*0.5:,.0f}',
                f'${cap*0.75:,.0f}',
                f'${cap:,.0f}+',
            ],
        ),
        annotations=[dict(
            text=note,
            xref='paper', yref='paper',
            x=0.5, y=-0.06,
            showarrow=False,
            font=dict(size=9, color='#777'),
            align='center',
        )],
        # Slow down the animation a bit
        updatemenus=[{
            'type': 'buttons',
            'showactive': False,
            'y': 0,
            'x': 0.5,
            'xanchor': 'center',
            'yanchor': 'top',
            'buttons': [{
                'label': '▶ Play',
                'method': 'animate',
                'args': [None, {
                    'frame': {'duration': 700, 'redraw': True},
                    'fromcurrent': True,
                    'transition': {'duration': 300},
                }],
            }, {
                'label': '⏸ Pause',
                'method': 'animate',
                'args': [[None], {
                    'frame': {'duration': 0, 'redraw': False},
                    'mode': 'immediate',
                    'transition': {'duration': 0},
                }],
            }],
        }],
    )

    # Slow frame transitions in the slider too
    for frame in fig.frames:
        frame['layout'] = {'transition': {'duration': 300}}

    out = os.path.join('Figures', fname)
    fig.write_html(out, include_plotlyjs='cdn')
    print(f"Saved -> {out}")


# ── Africa ────────────────────────────────────────────────────────────────────
build_gdp_map(
    africa_df,
    scope='africa',
    title='GDP per Capita Evolution: African Countries (1952–2007)',
    note=(
        'Color shows GDP per capita in USD. Darker red = wealthier. '
        'Click ▶ Play or drag the year slider to watch the evolution.'
    ),
    fname='ch3_map_gdp_africa_animated.html',
)

# ── Asia ──────────────────────────────────────────────────────────────────────
build_gdp_map(
    asia_df,
    scope='asia',
    title='GDP per Capita Evolution: Asian Countries (1952–2007)',
    note=(
        'Color shows GDP per capita in USD. Darker red = wealthier. '
        'Click ▶ Play or drag the year slider to watch the evolution. '
        'Note: Gulf states and Japan dominate the upper end.'
    ),
    fname='ch3_map_gdp_asia_animated.html',
)

print("Done. Both HTML files saved to Figures/")
```
